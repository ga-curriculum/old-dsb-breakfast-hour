```
# As a for loop

import math

def window(n, my_number):
  # change number to string - I need to iterate over it!
  my_number = str(my_number)
  products = []

  for pos, _ in enumerate(my_number):
    window = my_number[pos:pos+n]
    products.append(math.prod([int(i) for i in window]))

  return max(products)
```
