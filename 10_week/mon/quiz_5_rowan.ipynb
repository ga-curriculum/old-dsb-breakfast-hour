```python
%%time
import math

def window(n, my_number):
    # change my number to a string. I need to iterate over it
    my_number = str(my_number)
    products = []

    for pos, _ in enumerate(my_number):
        window = my_number[min(pos, len(my_number)-n): pos+n]
        products.append(math.prod([int(i) for i in window]))

    return max(products)
```
