## Quiz Qs

#### 1. Get the first 10 rows and all columns of the animals table (get all columns but limit the result of the query to 10 items).

Rubric (3 points possible):  
+1: Select all columns  
+1: From correct table  
+1: Limit results to 10  

```python
pd.read_sql("""

SELECT *
FROM animals
LIMIT 10

            """, conn)
```

#### 2. Get a list of the unique breeds of animals.

Rubric (3 points possible):  
+1: Select correct column  
+1: From correct table  
+1: Get unique values  

```python
pd.read_sql("""

SELECT DISTINCT breed
FROM animals;

            """, conn)
```

#### 3. Get all of the ages and weights of black dogs (`type` is `dog` and `color` is `black`).

Rubric (4 points possible):  
+1: Select correct columns  
+1: From correct table  
+1: Filter to just dogs  
+1: Filter to just black colors  

```python
pd.read_sql("""

SELECT age, weight
FROM animals
WHERE type = 'dog' AND color = 'black';

            """, conn)
```

#### 4. Get all of the types and breeds of animals that weigh more than 55 pounds.

Rubric (3 points possible):  
+1: Select correct columns  
+1: From correct table  
+1: Correct where statement  

```python
pd.read_sql("""

SELECT type, breed
FROM animals
WHERE weight > 55;

            """, conn)
```

#### 5. Get a list of all customers using gmail (customers whose emails end in `@gmail.com`).

Rubric (3 points possible):  
+1: Select correct columns  
+1: From correct table  
+1: Correct where statement  

```python
pd.read_sql("""

SELECT *
FROM customers
WHERE email LIKE '%@gmail.com'

            """, conn)
```

#### 6. Get the average weight by breed of breeds that average at least 40 pounds.

Rubric (4 points possible):  
+1: Select aggregate of weight  
+1: From correct table  
+1: Correct grouping  
+1: Only including breeds that average at least 40 pounds 

```python
pd.read_sql("""

SELECT breed, AVG(weight)
FROM animals
GROUP BY breed
HAVING AVG(weight) >= 40;

            """, conn)
```

#### 7. Get the last name of all people who have adopted a cat (`type` is `cat`).

Rubric (4 points possible):  
+1: Select correct column  
+1: From correct table  
+1: Correct join  
+1: Correct where statement  

```python
pd.read_sql("""

SELECT c.last_name
FROM customers c
INNER JOIN animals a ON a.id = c.pet_id
WHERE LOWER(a.type) = 'cat';

            """, conn)
```

#### 8. Grouping by type, get the average age of each type of animal. Name the average age column `avg_age`.

Rubric (4 points possible):  
+1: Select correct column  
+1: Correct re-naming of column  
+1: From correct table  
+1: Correct grouping  



```python
pd.read_sql("""

SELECT type, AVG(age) AS avg_age
FROM animals
GROUP BY type;

            """, conn)
```

#### 9. How many rabbits (`type` is `rabbit`) are located in New York (`state` is `New York`)?

Rubric (5 points possible):  
+1: Select count  
+1: From correct table  
+1: Correct join  
+1: Correct where statement for type  
+1: Correct where statement for state 

```python
pd.read_sql("""

SELECT COUNT(*)
FROM animals a
LEFT JOIN locations l on a.location_id = l.id
WHERE a.type = 'rabbit' AND l.state = 'New York';

            """, conn)
```

#### 10. Get the pet ID, breed, customer ID, and city of dogs (`type` is `dog`) who have been adopted (`adopted` is `Yes` or `yes`). Order the results alphabetically by breed.

Rubric (7 points possible):  
+1: Select correct columns  
+1: From correct table  
+1: Correct first join  
+1: Correct second join  
+1: Correct where statement for type  
+1: Correct where statement for adopted  
+1: Correct ordering  



```python
pd.read_sql("""

SELECT a.id, a.breed, c.customer_id, l.city
FROM animals a
LEFT JOIN customers c on a.id = c.pet_id 
LEFT JOIN locations l on a.location_id = l.id
WHERE LOWER(a.type) = 'dog' AND LOWER(adopted) = 'yes'
ORDER BY a.breed;

            """, conn)

# various arguments can be made about innder vs. left join here
# defend your choice
```
