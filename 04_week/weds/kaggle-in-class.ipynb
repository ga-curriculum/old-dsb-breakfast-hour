# **PART 2: Simple Linear Regression Model**

![train](./images/train.png)

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

Keep in mind that for this project, it will require a lot of cleaning, one hot encoding, and feature engineering (more to come later today!). You will typically be using more than one feature to make predictions (Multiple Linear Regression), but for the sake of this practice submission we will practice using just one feature! Which one should we pick?

```python
train['Overall Qual']
```

### STEP 1: Create your model

```python
# Set up X and y
X = train[['Overall Qual']]
y = train['SalePrice']
```

```python
# Train Test Split - Let's see how our model performs on unseen data before making predictions on the test.csv (df_test)!
X_train, X_val, y_train, y_val = train_test_split(X, y, random_state = 42)
```

* I'm not allowed to "know" information about the test set yet!

```python
# Preprocessing 
from sklearn.impute import SimpleImputer
```

```python
si = SimpleImputer()

# fit on my training set!
X_train_si = si.fit_transform(X_train)

# transform testing set (val set)
X_val_si = si.transform(X_val)

# Nothing changed in this case... We don't have any missing values
```

More preprocessing?
* fit and transform on training set
* transform on testing set

```python
# Instantiate 
lr = LinearRegression()
```

```python
# Fit on training data!
lr.fit(X_train_si, y_train)
```

### STEP 2: Evaluate on training and validation sets

![eval](./images/eval.png)

```python
# Train Score
lr.score(X_train_si, y_train)
```

```python
# Test Score
lr.score(X_val_si, y_test)
```

![train](./images/test.png)

### STEP 3: Okay COOL! Now that we have seen how our model would perform on unseen data, let's refit using all data!

```python
# Refit using all of the data in train.csv
lr.fit(X,y)

# If I actually had missing values - I'd need to impute!
```

### STEP 4: Make sale price predictions on homes that were in my `test.csv`

```python
preds = lr.predict(test[['Overall Qual']])
preds[:10]
```

### STEP 5: Create `SalePrice` column in df_test

```python
predictions = test.copy()

predictions['SalePrice'] = preds
```

### STEP 6: Create DataFrame with only `ID` and `SalePrice` columns from df_test
